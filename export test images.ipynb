```python
import torch, torchvision.utils as vutils, os

DS  = "cifar10"                       # cifar10 | svhn
ADV = "FGSM"                          # FGSM | BIM | DeepFool | CWL2  
N   = 40

d = f"/kaggle/input/datasets/sealeopard/attacked-pth-files/{DS.upper()}/{ADV}"
tag = f"resnet_{DS}_{ADV}"
clean = torch.load(f"{d}/clean_data_{tag}.pth", map_location="cpu")
adv   = torch.load(f"{d}/adv_data_{tag}.pth",   map_location="cpu")
noisy = torch.load(f"{d}/noisy_data_{tag}.pth", map_location="cpu")

mean = torch.tensor([0.4914, 0.4822, 0.4465]).view(1,3,1,1)
std  = torch.tensor([0.2023, 0.1994, 0.2010]).view(1,3,1,1)
def to_png(x): return torch.clamp(x*std + mean, 0, 1)

out = f"/kaggle/working/test_imgs/{DS}_{ADV}"
os.makedirs(out, exist_ok=True)
for i in range(N):
    vutils.save_image(to_png(clean[i:i+1]), f"{out}/clean_{i}.png")
    vutils.save_image(to_png(adv[i:i+1]),   f"{out}/adv_{i}.png")
    vutils.save_image(to_png(noisy[i:i+1]), f"{out}/noisy_{i}.png")
print(f"saved {N} clean + {N} adv + {N} noisy to {out}")
```
